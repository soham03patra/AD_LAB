Q1 Find out names of employees and their names of the department in which they are working

```python
out = pd.merge(emp, dept, left_on="department_id", right_on="dept_id",how="inner")\
[["last_name","department_name"]]
out.head(n=10)
```

Q2.For every department(Name), fetch the average salary in each department 

```python
df1 = emp[["department_id","salary"]].groupby("department_id").\
mean().rename(columns={"salary":"dept_avg_salary"})
df1
```

```python
pd.merge(emp, dept, left_on="department_id", right_on="dept_id",how="inner")\
[["department_name","salary"]].groupby("department_name").mean().\
rename(columns={"salary":"dept_avg_salary"}).reset_index()
```

```python
out = pd.merge(emp, dept, left_on="department_id", right_on="dept_id", how="right")\
[["last_name","department_name"]]
out
```
